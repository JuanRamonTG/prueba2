```python
#1.	Desarrolle un algoritmo en Python que calcule el IMPUESTO
# A LAS ACTIVIDADES ECONOMICAS de acuerdo con la siguiente tabla. El monto de la actividad económica
# debe ser introducido por el usuario. (40%)

def calcular_impuesto_actividad_economica(monto_actividad: float):
    if monto_actividad <= 0:
        raise ValueError("Por favor, ingresa un monto de actividad económica válido.")

    impuesto = 0
    if 0.01 <= monto_actividad <= 500:
        impuesto = 1.5
    elif 500.01 <= monto_actividad <= 1000:
        impuesto = 1.5 + (monto_actividad - 500.01) / 1000 * 3
    elif 1000.01 <= monto_actividad <= 2000:
        impuesto = 3 + (monto_actividad - 1000.01) / 1000 * 3
    elif 2000.01 <= monto_actividad <= 3000:
        impuesto = 6 + (monto_actividad - 2000.01) / 1000 * 3
    elif 3000.01 <= monto_actividad <= 6000:
        impuesto = 9 + (monto_actividad - 3000.01) / 1000 * 2
    elif 8000.01 <= monto_actividad <= 18000:
        impuesto = 15 + (monto_actividad - 8000.01) / 1000 * 2
    elif 18000.01 <= monto_actividad <= 30000:
        impuesto = 39 + (monto_actividad - 18000.01) / 1000 * 2
    elif 30000.01 <= monto_actividad <= 60000:
        impuesto = 63 + (monto_actividad - 30000.01) / 1000 * 1
    elif 60000.01 <= monto_actividad <= 100000:
        impuesto = 93 + (monto_actividad - 60000.01) / 1000 * 0.8
    elif 100000.01 <= monto_actividad <= 200000:
        impuesto = 125 + (monto_actividad - 100000.01) / 1000 * 0.7
    elif 200000.01 <= monto_actividad <= 300000:
        impuesto = 195 + (monto_actividad - 200000.01) / 1000 * 0.6
    elif 300000.01 <= monto_actividad <= 400000:
        impuesto = 255 + (monto_actividad - 300000.01) / 1000 * 0.45
    elif 400000.01 <= monto_actividad <= 500000:
        impuesto = 300 + (monto_actividad - 400000.01) / 1000 * 0.4
    elif 500000.01 <= monto_actividad <= 1000000:
        impuesto = 340 + (monto_actividad - 500000.01) / 1000 * 0.30
    elif 1000000.01 <= monto_actividad <= 99999999:
        impuesto = 490 + (monto_actividad - 1000000.01) / 1000 * 0.18


    return {
        "Monto de Actividad Económica": round(monto_actividad, 2),
        "Impuesto a Pagar": round(impuesto, 2)
    }


#
if __name__ == "__main__":
    try:
        monto = float(input("Ingresa el monto de la actividad económica: "))
        resultados = calcular_impuesto_actividad_economica(monto)

        print("\n--- Resultados ---")
        for clave, valor in resultados.items():
            print(f"{clave}: ${valor}")
    except ValueError as e:
        print(e)
```

```python
conversion_factors = {
    "Metro Cuadrado": 1.0,
    "Pie Cuadrado": 0.092903,
    "Vara Cuadrada": 0.698897,
    "Yarda Cuadrada": 0.836127,
    "Tarea": 437.5,
    "Manzana": 7000.0,
    "Hectárea": 10000.0
}

def convertir_area(valor, unidad_origen, unidad_destino):
    """Convierte un valor de un unidad de área a otra."""
    if unidad_origen not in conversion_factors or unidad_destino not in conversion_factors:
        return "Unidad de origen o destino no válida."

    valor_en_metros_cuadrados = valor * conversion_factors[unidad_origen]
    valor_convertido = valor_en_metros_cuadrados / conversion_factors[unidad_destino]

    return valor_convertido

if __name__ == "__main__":
    try:
        valor = float(input("Ingrese cantidad a convertir: "))

        print("\nUnidades disponibles:")
        unidades_list = list(conversion_factors.keys())
        for i, unidad in enumerate(unidades_list):
            print(f"{i+1}. {unidad}")

        while True:
            try:
                origen_index = int(input(f"Seleccione la unidad de origen (1-{len(unidades_list)}): ")) - 1
                if 0 <= origen_index < len(unidades_list):
                    unidad_origen = unidades_list[origen_index]
                    break
                else:
                    print("Selección no válida. Intente de nuevo.")
            except ValueError:
                print("Entrada no válida. Por favor, ingrese un número.")

        while True:
            try:
                destino_index = int(input(f"Seleccione la unidad de destino (1-{len(unidades_list)}): ")) - 1
                if 0 <= destino_index < len(unidades_list):
                    unidad_destino = unidades_list[destino_index]
                    break
                else:
                    print("Selección no válida. Intente de nuevo.")
            except ValueError:
                print("Entrada no válida. Por favor, ingrese un número.")

        resultado = convertir_area(valor, unidad_origen, unidad_destino)

        if isinstance(resultado, str):
            print(resultado)
        else:
            print(f"{valor} {unidad_origen} es igual a {resultado:.2f} {unidad_destino}")

    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número para el valor.")
```

```python
def simular_cajero(cantidad_a_retirar: float):
    if cantidad_a_retirar <= 0:
        print("Por favor, ingrese una cantidad válida mayor a cero.")
        return

    print(f"Cantidad a retirar: {cantidad_a_retirar:.2f}")
    print("\nDesglose:")

    billetes = {
        20: 0,
        10: 0,
        5: 0,
        1: 0
    }
    monedas = {
        0.25: 0,
        0.10: 0,
        0.05: 0,
        0.01: 0
    }

    cantidad_en_centavos = int(round(cantidad_a_retirar * 100))

    # Calcular billetes
    for valor_billete in sorted(billetes.keys(), reverse=True):
        num_billetes = cantidad_en_centavos // (valor_billete * 100)
        if num_billetes > 0:
            billetes[valor_billete] = num_billetes
            cantidad_en_centavos %= (valor_billete * 100)

    for valor_moneda in sorted(monedas.keys(), reverse=True):
        num_monedas = cantidad_en_centavos // int(round(valor_moneda * 100))
        if num_monedas > 0:
            monedas[valor_moneda] = num_monedas
            cantidad_en_centavos %= int(round(valor_moneda * 100))


    for valor_billete, num_billetes in billetes.items():
        if num_billetes > 0:
            print(f"{num_billetes} billete{'s' if num_billetes > 1 else ''} de a {valor_billete}")

    for valor_moneda, num_monedas in monedas.items():
        if num_monedas > 0:
            print(f"{num_monedas} moneda{'s' if num_monedas > 1 else ''} de {valor_moneda:.2f} ctvs.")

if __name__ == "__main__":
    try:
        cantidad = float(input("Ingrese la cantidad a retirar: "))
        simular_cajero(cantidad)
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número.")
```
