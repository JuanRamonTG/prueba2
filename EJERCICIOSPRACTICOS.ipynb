```python
def calcular_salario(salario: float):
    if salario <= 0:
        raise ValueError("Por favor, ingresa un salario válido.")

    # ISSS (3% con tope de $30)
    isss = min(salario * 0.03, 30)

    # AFP (6.25%)
    afp = salario * 0.0625

    # Sueldo gravado
    sueldo_gravado = salario - isss - afp

    # ISR
    renta = 0
    if sueldo_gravado > 2038.1:
        renta = (sueldo_gravado - 2038.1) * 0.30 + 288.57
    elif sueldo_gravado > 895.24:
        renta = (sueldo_gravado - 895.24) * 0.20 + 60
    elif sueldo_gravado > 472:
        renta = (sueldo_gravado - 472) * 0.10 + 17.67

    # Totales
    descuentos_totales = isss + afp + renta
    salario_neto = salario - descuentos_totales

    return {
        "Salario Base": round(salario, 2),
        "ISSS": round(isss, 2),
        "AFP": round(afp, 2),
        "Renta": round(renta, 2),
        "Descuentos Totales": round(descuentos_totales, 2),
        "Salario Neto": round(salario_neto, 2),
        "Sueldo Gravado": round(sueldo_gravado, 2)
    }


#
if __name__ == "__main__":
    try:
        salario = float(input("Ingresa tu salario: "))
        resultados = calcular_salario(salario)

        print("\n--- Resultados ---")
        for clave, valor in resultados.items():
            print(f"{clave}: ${valor}")
    except ValueError as e:
        print(e)
```

```python
def calcular_pago(metros):
    cuota_fija = 6
    if metros <= 18:
        return cuota_fija
    elif metros <= 28:
        exceso = metros - 18
        return cuota_fija + exceso * 0.45
    else:
        exceso_19_28 = 10 * 0.45  # de 19 a 28 siempre serán 10 m³
        exceso_29 = (metros - 28) * 0.65
        return cuota_fija + exceso_19_28 + exceso_29


print("Calculadora de pago de agua potable")
metros = float(input("Ingrese los metros cúbicos consumidos: "))

pago = calcular_pago(metros)

print(f"El valor a pagar por {metros} m³ consumidos es: ${pago:.2f}")
```

```python

"""
Conversores
"""

# =========================
# Motor de conversión genérico
def convert(value: float, from_u: str, to_u: str, table: dict) -> float:
    if from_u not in table:
        raise KeyError(f"Unidad de origen no soportada: {from_u}")
    if to_u not in table:
        raise KeyError(f"Unidad de destino no soportada: {to_u}")
    base_value = value * table[from_u]   # convertir a base
    return base_value / table[to_u]      # convertir a destino


# =========================
# Tablas de unidades
LONGITUD = {
    "nm": 1e-9, "µm": 1e-6, "mm": 1e-3, "cm": 1e-2,
    "m": 1.0, "km": 1e3, "in": 0.0254, "ft": 0.3048,
    "yd": 0.9144, "mi": 1609.344, "nmi": 1852.0,
}
MASA = {
    "µg": 1e-9, "mg": 1e-6, "g": 1e-3, "kg": 1.0,
    "t": 1e3, "oz": 0.0283495, "lb": 0.453592,
    "st": 6.35029, "gr": 0.0000648, "ct": 0.0002,
}
ALMACENAMIENTO = {
    "b": 1/8.0, "B": 1.0, "kB": 1000.0, "MB": 1e6,
    "GB": 1e9, "TB": 1e12, "KiB": 1024.0, "MiB": 1024.0**2,
    "GiB": 1024.0**3, "TiB": 1024.0**4,
}
TIEMPO = {
    "ns": 1e-9, "µs": 1e-6, "ms": 1e-3, "s": 1.0,
    "min": 60.0, "h": 3600.0, "d": 86400.0, "wk": 604800.0,
    "mo": 2629800.0, "yr": 31557600.0,
}
VOLUMEN = {
    "mL": 1e-3, "L": 1.0, "m3": 1000.0, "in3": 0.0163871,
    "ft3": 28.3168, "tsp": 0.00492892, "tbsp": 0.0147868,
    "fl_oz": 0.0295735, "cup": 0.236588, "pt": 0.473176,
    "qt": 0.946353, "gal": 3.78541,
}
AREA = {
    "mm2": 1e-6, "cm2": 1e-4, "m2": 1.0, "ha": 10000.0,
    "km2": 1e6, "in2": 0.00064516, "ft2": 0.092903,
    "yd2": 0.836127, "ac": 4046.86, "mi2": 2.59e6,
}
ENERGIA = {
    "J": 1.0, "kJ": 1e3, "cal": 4.184, "kcal": 4184.0,
    "Wh": 3600.0, "kWh": 3.6e6, "BTU": 1055.06, "eV": 1.6e-19,
}
VELOCIDAD = {
    "m/s": 1.0, "km/h": 1000.0/3600.0, "mph": 1609.344/3600.0,
    "kn": 1852.0/3600.0, "ft/s": 0.3048, "c": 299792458.0,
}
PRESION = {
    "Pa": 1.0, "kPa": 1e3, "MPa": 1e6, "bar": 1e5,
    "atm": 101325.0, "Torr": 133.322, "psi": 6894.76,
}

# Registro de categorías
UNITS = {
    "monedas": monedas_con_equivalencias,
    "longitud": LONGITUD,
    "masa": MASA,
    "almacenamiento": ALMACENAMIENTO,
    "tiempo": TIEMPO,
    "volumen": VOLUMEN,
    "área": AREA,
    "energía": ENERGIA,
    "velocidad": VELOCIDAD,
    "presión": PRESION,
}


# Conversor de monedas
def convert_currency():
    monedas_con_equivalencias = {
        "usd": 1,
        "mxn": 18.8284,
        "eur": 0.856,
        "gbp": 0.783,
        "jpy": 151.455,
        "cny": 7.196,
        "cad": 1.3,
        "aud": 1.45,
        "chf": 0.87,
        "sek": 10.5,
        "brl": 5.2,
        "inr": 83.0,
        "btc": 0.000088
    }

    print("\nMonedas disponibles:")
    for moneda in monedas_con_equivalencias:
        print(moneda.upper())

    nombre_moneda_origen = ""
    while nombre_moneda_origen.lower() not in monedas_con_equivalencias:
        nombre_moneda_origen = input("Moneda origen: ").strip().lower()

    nombre_moneda_destino = ""
    while nombre_moneda_destino.lower() not in monedas_con_equivalencias:
        nombre_moneda_destino = input("Moneda destino: ").strip().lower()

    cantidad = float(input(f"Ingresa la cantidad de {nombre_moneda_origen.upper()}: "))

    equivalencia_origen = monedas_con_equivalencias[nombre_moneda_origen]
    equivalencia_destino = monedas_con_equivalencias[nombre_moneda_destino]
    resultado = (equivalencia_destino / equivalencia_origen) * cantidad

    print(f"\n{cantidad:.2f} {nombre_moneda_origen.upper()} equivale a {resultado:.2f} {nombre_moneda_destino.upper()}.")



# Menú principal
def run_converter():
    while True:
        print("=== Conversores Universales ===")
        categorias = list(UNITS.keys())
        for i, c in enumerate(categorias, start=1):
            print(f"{i}. {c}")
        print("0. Salir")

        choice = int(input("Elige categoría (número): "))
        if choice == 0:
            print("Adiós")
            break

        categoria = categorias[choice-1]

        if categoria == "monedas":
            convert_currency()
        else:
            table = UNITS[categoria]
            print(f"\nElegiste: {categoria}")
            print("Unidades disponibles:", ", ".join(table.keys()))
            value = float(input("Cantidad: "))
            from_u = input("Desde (unidad): ").strip()
            to_u = input("Hacia (unidad): ").strip()

            try:
                result = convert(value, from_u, to_u, table)
                print(f"\n {value} {from_u} = {result} {to_u}\n")
            except Exception as e:
                print("Error:", e)

run_converter()
```

```python

"""
Conversor de unidades
"""

# Tabla de conversiones
conversiones = {
    "Caja-Unidad": {"factor": 25, "desde": "Caja", "hasta": "Unidad"},
    "Saco-Libras": {"factor": 50, "desde": "Saco", "hasta": "Libras"},
    "Fardo-Unidad": {"factor": 5, "desde": "Fardo", "hasta": "Unidad"},
}

def convertir(valor, factor, direccion="normal"):

    if direccion == "normal":
        return valor * factor
    elif direccion == "inversa":
        return valor / factor
    else:
        raise ValueError("Dirección de conversión no válida")

def mostrar_menu():
    print("\n--- Conversor de Unidades ---")
    print("1. Caja ↔ Unidad")
    print("2. Saco ↔ Libras")
    print("3. Fardo ↔ Unidad")
    print("0. Salir")

while True:
    mostrar_menu()
    opcion = input("Selecciona una opción: ")

    if opcion == "0":
        print("Saliendo del conversor...")
        break
    elif opcion in ["1", "2", "3"]:
        # Elegir la conversión
        if opcion == "1":
            key = "Caja-Unidad"
        elif opcion == "2":
            key = "Saco-Libras"
        elif opcion == "3":
            key = "Fardo-Unidad"

        direccion = input("¿Deseas convertir de {} a {} (n) o de {} a {} (i)? ".format(
            conversiones[key]["desde"], conversiones[key]["hasta"],
            conversiones[key]["hasta"], conversiones[key]["desde"]
        )).lower()

        valor = float(input("Ingresa el valor a convertir: "))

        if direccion == "n":
            resultado = convertir(valor, conversiones[key]["factor"], "normal")
            print(f"{valor} {conversiones[key]['desde']} = {resultado} {conversiones[key]['hasta']}")
        elif direccion == "i":
            resultado = convertir(valor, conversiones[key]["factor"], "inversa")
            print(f"{valor} {conversiones[key]['hasta']} = {resultado} {conversiones[key]['desde']}")
        else:
            print("Opción de dirección no válida. Intenta de nuevo.")
    else:
        print("Opción no válida. Intenta de nuevo.")
```
