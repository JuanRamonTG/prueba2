```python
# Función con parámetro
def mostrar_flujo(nombre):
  print("Inicio de la función")
  print(f"El parámetro 'nombre' recibe el valor: {nombre}") # <-- aquí el parámetro ya tiene el argumento

  mensaje = f"Hola, {nombre}!" # variable normal usando el parámetro
  print("Se crea la variable 'mensaje' con el valor:", mensaje)

  print("Fin de la función\n")

# Llamadas a la función con distintos argumentos
mostrar_flujo("Ana")
mostrar_flujo("Luis")
mostrar_flujo("María")
```
