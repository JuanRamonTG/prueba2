```python
#reduce
from functools import reduce

numeros = [5,4,9,6,8,2,3]
suma = reduce(lambda acum, num: num+acum, numeros)

sumaPares = reduce(lambda acum, num: num+acum, filter(lambda num: num%2==0, numeros))
sumaPar = reduce(lambda acum, num: acum + num if num%2==0 else acum, numeros,0)

print("la suma es: ", suma)
print("la suma de los pares es: ", sumaPares)
print("Suma Pares: ", sumaPar)

a#gente manus
```

```python
class calculadora:
  def mayor(self, serie):
    nMayor = 0
    for i in serie:
      num = int(i)
      if(num>nMayor):
        nMayor = num
    return nMayor

class nNumeroMenor:
  def menor(self, serie):
    nMenor = 99999999999
    for i in serie:
      num = int(i)
      if(num<nMenor):
        nMenor = num
    return nMenor

```

```python
calcular = calculadora()
calcula = nNumeroMenor()
serie = input("Serie ejm: 1,5,4,9,3: ").split(",")
print(serie)
print("El Mayor es: ", calcular.mayor(serie))
print("Menor es: ", calcula.menor(serie))
```

```python
class calculadora:
  def mayor_menor(self, serie):
    return max(serie), min(serie)
```

```python
calcular = calculadora()
serie = input("Serie ejm: 1,5,4,9,3: ").split(",")
mayor, menor = calcular.mayor_menor(serie)
print("El Mayor es: ", mayor, "Menor es: ", menor)
```

```python
#Encontrar la persona de mayor edad en una lista de datos

personas = [
     {"nombre":"Luis", "edad": 30},
     {"nombre":"Juan", "edad": 40},
     {"nombre":"Jorge", "edad": 20},
     {"nombre":"Jose", "edad": 10},
]

persona_mayor_edad = max(personas, key=lambda x: x['edad'])

print("La persona de mayor edad es:", persona_mayor_edad["nombre"])
```
